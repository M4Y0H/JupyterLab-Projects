```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import joblib as jb
from tabulate import tabulate
from sklearn import tree

music_dataset = pd.read_excel(r'C:\Users\LAPTOP\OneDrive\Desktop\musics.xlsx')
gen_x = music_dataset.drop(columns=['genre'])
gen_y = music_dataset['genre']
gen_x_train, gen_x_test, gen_y_train, gen_y_test = train_test_split(gen_x, gen_y, test_size = 0.1)
header = ['genre'] 

module = DecisionTreeClassifier()
module.fit(gen_x_train, gen_y_train)
pred = module.predict(gen_x_test)
jb.dump(module, 'music-recommendation.joblib')
tree.export_graphviz(module, out_file = 'music-recommendation.dot'
                    feature_names = ['age', 'gender']
                    class_names = sorted(gen_y.unique())
                    label = 'all'
                    rounded = True
                    filled = True)

genre = pred
table = zip(genre)

print(gen_x_test)          
print(tabulate(table, headers=header, floatfmt='.4f'))
score = accuracy_score(gen_y_test, pred) 
print('accuracy score = ', score)
```
