```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn import linear_model
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import log_loss
```

```python
loc = "C:/Users/LAPTOP/Downloads/ChurnData.csv"
telecom_data = pd.read_csv(loc)
```

```python
telecom_data.sample(5)
```

```python
telecom_data  = telecom_data[['tenure', 'age', 'address', 'income', 'ed', 'employ', 'equip', 'churn']]
telecom_data['churn'] = telecom_data['churn'].astype(int)
telecom_data
```

```python
telecom_data.corr()
```

```python
ind_var = np.asarray(telecom_data[['tenure', 'age', 'address', 'income', 'ed', 'employ', 'equip']])
ind_var[0:5]
```

```python
dep_var = np.asarray(telecom_data[['churn']].values.ravel())
dep_var[0:5]
```

```python
ind_norm = StandardScaler().fit(ind_var).transform(ind_var)
ind_norm[0:5]
```

```python
train_ind, test_ind, train_dep, test_dep = train_test_split(ind_norm, dep_var, test_size=0.2, random_state=42)
```

```python
LogisticRegression = linear_model.LogisticRegression()
LogR = LogisticRegression.fit(train_ind, train_dep)
pred_dep = LogR.predict(test_ind)
pred_dep[11:21]
```

```python
prob_dep = LogR.predict_proba(test_ind)
prob_dep[11:21]
```

```python
test_dep[11:21]
```

```python
coef = pd.Series(LogR.coef_[0], index= telecom_data.columns[:-1])
coef.sort_values().plot(kind='barh')
plt.title("Feature Coefficients in Logistic Regression Churn Model")
plt.xlabel("Coefficient Value")
plt.show()
```

```python
log_loss(test_dep, prob_dep)
```
