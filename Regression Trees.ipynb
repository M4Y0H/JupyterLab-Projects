```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
import time
```

```python
path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ML0101EN-SkillsNetwork/labs/Module%203/data/real_estate_data.csv"
housin_data = pd.read_csv(path) 
housin_data.head()
housin_data.shape
```

```python
housin_data.isna().sum()
```

```python
housin_data.dropna(inplace=True)
housin_data.isna().sum()
```

```python
feature_x = housin_data.drop(columns=["MEDV"])
feature_y = housin_data["MEDV"]
feature_x.head()
```

```python
feature_y.head()
```

```python
train_x, test_x, train_y, test_y = train_test_split(feature_x, feature_y, test_size = 0.2, random_state = 4)
initial_time = time.time()
reg_tree = DecisionTreeRegressor(criterion = "friedman_mse").fit(train_x, train_y)
time_delta = time.time() - initial_time
print('Time taken to train model: {0:.4f}s'.format(time_delta))
```

```python
reg_tree.score(train_x, train_y)
```

```python
predict_y = reg_tree.predict(test_x)
print("$",(test_y - predict_y).abs().mean()*1000)
```

```python
from sklearn.metrics import r2_score
reg_tree_mae = DecisionTreeRegressor(criterion = 'absolute_error').fit(train_x, train_y)
pred_mae_y = reg_tree_mae.predict(test_x)
print("R2 Score: %.5f" % r2_score(test_y, pred_mae_y))
print("MAE: %.5f" % (test_y - pred_mae_y).abs().mean())
```
