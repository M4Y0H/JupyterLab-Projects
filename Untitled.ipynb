```python
import numpy as np

weights = np.around(np.random.uniform(size=6), decimals=2)
biases = np.around(np.random.uniform(size=3), decimals=2)
```

```python
print(weights)
print(biases)
```

```python
x_1 = 0.1
x_2 = 0.15

print('The inputs are {} and {}'.format(x_1, x_2))
```

```python
z_11 = x_1*weights[0] + x_2*weights[1] + biases[0]
z_12 = x_1*weights[2] + x_2*weights[3] + biases[1]

print('The weighted sum the inputs at the first node is {}'.format(z_11))
print('The weighted sum the inputs at the second node is {}'.format(z_12))
```

```python
a_11 = 1.0 / (1.0 + np.exp(-z_11))
a_12 = 1.0 / (1.0 + np.exp(-z_12))

print('The activation of the first node is {}'.format(np.around(a_11, decimals=4)))
print('The activation of the second node is {}'.format(np.around(a_12, decimals=4)))
```

```python
z_2 = a_11*weights[4] + a_12*weights[5] + biases[2]

print('The weighted sum of all nodes at the output layer is {}'.format(np.around(z_2, decimals=4)))
```

```python
a_2 = 1.0 / (1.0 + np.exp(-z_2))
a_2
```
