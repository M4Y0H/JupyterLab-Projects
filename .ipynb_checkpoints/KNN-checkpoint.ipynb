```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import pylab as py
from sklearn import preprocessing
```

```python
url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ML0101EN-SkillsNetwork/labs/Module%203/data/teleCust1000t.csv'
file = pd.read_csv(url)
file.head(100)
```

```python
file['custcat'].value_counts()
```

```python
file.hist(column = 'income', bins = 50)
```

```python
file.columns
```

```python
Xnet =file[['region', 'tenure', 'age', 'marital', 'address', 'income', 'ed',
       'employ', 'retire', 'gender', 'reside', 'custcat']].values
Xnet[0:5]
```

```python
y = file[['custcat']].values
y[0:200]
```

Xnet = preprocessing.StandardScaler().fit(Xnet).transform(Xnet.astype(float))
Xnet[0:4]

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(Xnet, y, test_size = 0.2, random_state = 4)
print('Train Set: ', x_train.shape,',',y_train.shape)
print('Test Set: ', x_test.shape,',',y_test.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier
k = 10
neighbours = KNeighborsClassifier(n_neighbors=k).fit(x_train, y_train)
neighbours
y_result = neighbours.predict(x_test)
y_result[0:10]
```

```python
from sklearn import metrics
print("Train set Accuracy: ", metrics.accuracy_score(y_train, neighbours.predict(x_train)))
print("Test set Accuracy: ", metrics.accuracy_score(y_test, y_result))
```

```python
k = 11
mean_acc = np.zeros((k-1))
std_acc = np.zeros((k-1))

for n in range(1,k): 
    neigh = KNeighborsClassifier(n_neighbors = n).fit(x_train,y_train)
    y_result=neigh.predict(x_test)
    mean_acc[n-1] = metrics.accuracy_score(y_test, y_result)

    std_acc[n-1]=np.std(y_result==y_test)/np.sqrt(y_result.shape[0])

mean_acc
```

```python
plt.plot(range(1,k),mean_acc,'g')
plt.fill_between(range(1,k),mean_acc - 1 * std_acc,mean_acc + 1 * std_acc, alpha=0.10)
plt.fill_between(range(1,k),mean_acc - 3 * std_acc,mean_acc + 3 * std_acc, alpha=0.10,color="green")
plt.legend(('Accuracy ', '+/- 1xstd','+/- 3xstd'))
plt.ylabel('Accuracy ')
plt.xlabel('Number of Neighbors (K)')
plt.tight_layout()
plt.show()
```
