```python
def collect_details():
    details = {}
    name = input("Hi, What is your name? ")
    if name:
        details["Name"] = name
        age = input("Nice! and How old are you? ")
        place = input("Also, before I forget where are you from? ")

        if age and place:
            details["Age"] = age
            details["Place of Birth"] = place
            print("Thank You, I have your deatils saved up")

        else: 
            print("Your other details please")

    else:
        print("Please input your name")
    
    return details

def share_details(**details):
    print("Your Details are: ")
    for key, value in details.items():
        print("{}: {}".format(key, value))

details = collect_details()

share_details(**details)
```
