```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import pylab as py
from sklearn.tree import DecisionTreeClassifier
import sklearn.tree as tree
```

```python
file_path = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ML0101EN-SkillsNetwork/labs/Module%203/data/drug200.csv'
drugs_data = pd.read_csv(file_path)
drugs_data.head(15)
```

```python
drug_val = drugs_data[["Age", "Sex", "BP", "Cholesterol", "Na_to_K"]].values
drug_val[0:5]
```

```python
drug_val = drugs_data[["Age", "Sex", "BP", "Cholesterol", "Na_to_K"]].values
from sklearn.preprocessing import LabelEncoder
le_sex = LabelEncoder()
le_sex.fit(["F", "M"])
drug_val[:, 1] = le_sex.transform(drug_val[:, 1])

le_BP = LabelEncoder()
le_BP.fit(["LOW", "NORMAL", "HIGH"])
drug_val[:, 2] = le_BP.transform(drug_val[:, 2])

le_Chol = LabelEncoder()
le_Chol.fit(["NORMAL", "HIGH"])
drug_val[:, 3] = le_Chol.transform(drug_val[:, 3])

drug_val[0:5]
```

```python
y_drug = drugs_data[["Drug"]]
y_drug[:]
```

```python
from sklearn.model_selection import train_test_split
train_x, test_x, train_y, test_y = train_test_split(drug_val, y_drug, test_size = 0.2, random_state = 4)
drugTree = DecisionTreeClassifier(criterion = "entropy", ).fit(train_x, train_y)
predDrug = drugTree.predict(test_x)
print(test_y[0:10])
print(predDrug[0:10])
```

```python
from sklearn import metrics
print("Decision Tree Accuracy: %.2f" % metrics.accuracy_score(test_y, predDrug))
```
