```python
%%writefile spiders.py
import scrapy

class BookSpider(scrapy.Spider):
    name = "books"
    start_urls = ["https://books.toscrape.com"]

    def parse(self, response):
        for title in response.css("h3 a::attr(title)").getall():
            yield {"title": title}
```
