```python
import smtplib
from email.mime.text import MIMEText
from email.mime.multipart import MIMEMultipart
```

```python
msg =  MIMEText("Hello, Customer.")
msg["Sub"] = "Local Host"
msg["From"] = "demo@email.com"
msg["To"] = "local@email.com"

with smtplib.SMTP("localhost", 1025) as server:
    server.sendmail("demo@email.com", "local@email.com", msg.as_string())
```
