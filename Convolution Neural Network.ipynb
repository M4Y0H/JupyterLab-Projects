```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import keras
```

```python
from keras.models import Sequential
from keras.layers import Input, Dense, Conv2D, MaxPool2D, AveragePooling2D, Flatten
from keras.utils import to_categorical
```

```python
from keras.datasets import mnist

(in_tr, out_tr), (in_te, out_te) = mnist.load_data()

in_tr = in_tr.reshape(in_tr.shape[0], 28, 28, 1)
in_te = in_te.reshape(in_te.shape[0], 28, 28, 1)
```

```python
def cnn():
    model = Sequential()
    input_shape = in_tr.shape[1:]
    model.add(Input(shape=(input_shape)))
    model.add(Conv2D(16, kernel_size=(3,3), strides=(1,1), activation='relu'))
    model.add(MaxPool2D(pool_size=(2,2), strides=(2,2)))
    model.add(Conv2D(32, kernel_size=(3,3), strides=(1,1), activation='relu'))
    model.add(AveragePooling2D(pool_size=(2,2), strides=(1,1)))
    model.add(Conv2D(64, kernel_size=(3,3), strides=(1,1), activation='relu'))
    model.add(MaxPool2D(pool_size=(2,2), strides=(2,2)))
    model.add(Flatten())
    model.add(Dense(86, activation='relu'))
    model.add(Dense(n_class, activation='softmax'))

    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    return model
```

```python
in_tr = in_tr/255
in_te = in_te/255

out_tr = to_categorical(out_tr)
out_te = to_categorical(out_te)

n_class = out_tr.shape[1]
```

```python
con_network = cnn()

con_network.fit(in_tr, out_tr, validation_split=0.2, epochs=50, batch_size=150, verbose=2)
score = con_network.evaluate(in_te, out_te, verbose=2)
```

```python
print("Accuracy: {} \n Error: {}".format(score[1], 100-score[1]*100))
```

```python
from keras.datasets import fashion_mnist
```

```python
(one_in, one_out), (two_in, two_out) = fashion_mnist.load_data()
one_in = one_in.reshape(one_in.shape[0], 28, 28, 1)
two_in = two_in.reshape(two_in.shape[0], 28, 28, 1)

one_in = one_in/255
two_in = two_in/255
one_out = to_categorical(one_out) 
two_out = to_categorical(two_out)

n_cls = len(one_out[1])
```

```python
def clothing_cnn():
    model = Sequential()
    input_shape = one_in.shape[1:]
    model.add(Input(shape=input_shape))
    model.add(Conv2D(12, kernel_size=(3,3), strides=(1,1), activation='relu'))
    model.add(MaxPool2D(pool_size=(2,2), strides=(2,1)))
    model.add(Conv2D(24, kernel_size=(5,5), strides=(1,1), activation='relu'))
    model.add(MaxPool2D(pool_size=(2,2), strides=(1,2)))
    model.add(Conv2D(48, kernel_size=(5,5), strides=(1,1), activation='relu'))
    model.add(MaxPool2D(pool_size=(2,2), strides=(2,2)))
    model.add(Flatten())

    model.add(Dense(64, activation='relu'))
    model.add(Dense(n_cls, activation='softmax'))

    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    return model
```

```python
clothing_classifier = clothing_cnn()

clothing_classifier.fit(one_in, one_out, validation_split=.15, epochs=50, batch_size=86, verbose=2)

met = clothing_classifier.evaluate(two_in, two_out, verbose=2)
```
