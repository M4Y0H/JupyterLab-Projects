```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
%matplotlib inline 
```

```python
path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-ML0101EN-SkillsNetwork/labs/Module%202/data/FuelConsumptionCo2.csv"
```

```python
multi_file = pd.read_csv(path)
multi_file.sample(1)
```

```python
multi_file.describe()
```

```python
multi_file = multi_file.drop(["MODELYEAR", "MAKE", "MODEL", "VEHICLECLASS", "TRANSMISSION", "FUELTYPE"], axis=1)
```

```python
multi_file.corr()
```

```python
multi_file = multi_file.drop(["CYLINDERS", "FUELCONSUMPTION_CITY", "FUELCONSUMPTION_HWY", "FUELCONSUMPTION_COMB"], axis=1)
```

```python
multi_file.sample(10)
```

```python
matrix = pd.plotting.scatter_matrix(multi_file, alpha=0.35)
for mat in matrix.flatten():
    mat.xaxis.label.set_rotation(90)
    mat.yaxis.label.set_rotation(0)
    mat.yaxis.label.set_ha('right')
plt.tight_layout()
plt.gcf().subplots_adjust(wspace=0, hspace=0)
plt.show()
```
