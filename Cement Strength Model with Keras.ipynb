```python
con_df = pd.read_csv("C:/Users/User/Downloads/concrete_data.csv")
con_df
```

```python
con_df.shape
```

```python
predictors = con_df.drop(columns="Strength")
target = con_df["Strength"]
```

```python
norm_predicts = (predictors - predictors.mean())/predictors.std()
norm_predicts
```

```python
in_num = norm_predicts.shape[1]
```

```python
from keras.models import Sequential
from keras.layers import Dense
from keras.layers import Input
```

```python
def regression_model():
    model = Sequential()
    model.add(Input(shape=(in_num,)))
    model.add(Dense(256, activation='relu'))
    model.add(Dense(80, activation='relu'))
    model.add(Dense(42, activation='relu'))
    model.add(Dense(1))

    model.compile(optimizer='adam', loss='mean_squared_error')
    return model
```

```python
concrete_model = regression_model()
```

```python
concrete_model.fit(norm_predicts, target, validation_split=0.3, verbose=2, epochs=100)
```

```python
def regression_model_upt():
    model = Sequential()
    model.add(Input(shape=(in_num,)))
    model.add(Dense(128, activation='relu'))
    model.add(Dense(50, activation='relu'))
    model.add(Dense(25, activation='relu'))
    model.add(Dense(36, activation='relu'))
    model.add(Dense(25, activation='relu'))
    model.add(Dense(1))

    model.compile(optimizer='adam', loss='mean_squared_error')
    return model
```

```python
model = regression_model_upt()
```

```python
model.fit(norm_predicts, target, validation_split=0.10, verbose=1, validation_batch_size=50, epochs=100)
```

```python
sample = norm_predicts.sample(1)
sample_in = sample.index[0]
pred_y = model.predict(sample)
```
