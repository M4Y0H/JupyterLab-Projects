```python
import requests

url = "https://quotes.toscrape.com/"
response = requests.get(url)

print("Status Code: {}".format(response.status_code))
print("Preview of HTML:\n")
print(response.text[:1000])
```

```python
from bs4 import BeautifulSoup

beausoup = BeautifulSoup(response.text, "html.parser")

print(beausoup.title)
print(beausoup.title.text)
```

```python
quotes = beausoup.find_all("span", class_="text")
authors = beausoup.find_all("small", class_="author")

for q, a in zip(quotes, authors):
    print("{} - {}".format(q.text, a.text))
```

```python
links = [a["href"] for a in beausoup.find_all("a", href=True)]

for l in links:
    print(l)
```

```python
all_q = []

for page in range(1,5):
    
    url = "https://quotes.toscrape.com/page/{}/".format(page)
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")
    quotes = soup.find_all("span", class_="text")
    authors = soup.find_all("small", class_="author")
    for a, q in zip(quotes, authors):
        all_q.append("{} - {}".format(a.text,q.text))
```

```python
import pandas as pd
split = [item.split(" - ", 1) for item in all_q]
quote_df = pd.DataFrame(split, columns=["Quotes", "Authors"])
```

```python
quote_df.to_csv("Great_Quotes.csv", index=False)
```
