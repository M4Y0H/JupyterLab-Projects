```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
data = {
    'Name': ['Alan', 'Bob', 'Charles', 'Dave', 'Even', 'Guan', 'Heller', 'Ivan', 'John', 'Franky'],
    'Hrs_St': [2, 3, 4, 5, 6, 7, 8, 9, 5, 10],
    'Score': [32, 44, 56, 64, 71, 83, None, 92, 97, 55]
}

df = pd.DataFrame(data)
df.to_csv("students.csv", index=False)

df.head()
```

```python
df.info()
```

```python
df['Score'] = df['Score'].fillna(df['Score'].mean())

df.rename(columns={'Hrs_St':'Hours'}, inplace=True)
df['Pass'] = df['Score'].apply(lambda x: True if x >= 50 else False)
```

```python
 df_corr = df.drop(columns=['Name'])
```

```python
plt.scatter(df['Hours'], df['Score'])
plt.xlabel('Hours Spent Studing')
plt.ylabel('Test Scores')
plt.title('Hours vs Score')
plt.show()

sb.heatmap(df_corr.corr(), annot=True, cmap='coolwarm')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

x = df[['Hours']]
y = df[['Score']]
model = LinearRegression()
model.fit(x, y)

prediction = model.predict([[5]])
print("Prediction Score for 5 Hours of Study {}".format(prediction))
```

```python
import pandas as pd
data = {
    "Name": ["Kuku", "Gbemi", "Alex", "Emma", "Ileri"],
    "Age" : [24, 26, 23, 22, 25],
    "City": ["Oyo", "Lagos", "Osun", "Abuja", "Osun"],
    "Salary": [50000, 120000, 100000, 230000, 100000]
}

df = pd.DataFrame(data)
df
```

```python
df["Department"] = ["IT", "HR", "IT", "Finance", "HR"]
```

```python
df
```

```python
df.groupby("Department")["Salary"].mean()
```

```python
df.groupby("Department")["Age"].mean()
```

```python
agg_data = df.groupby("Department").agg({
    "Salary": ["mean", "max", "min", "std", "var", "sum"],
    "Age": ["mean", "count"]
})

agg_data[("Salary", "std")] = agg_data[("Salary", "std")].round(2)
agg_data[("Salary", "var")] = agg_data[("Salary", "var")].round(2)
print("Aggregate Values by Department: \n {}".format(agg_data))
```

```python
multi_agg_data = df.groupby(["Department", "City"]).agg({
    "Salary": ["mean", "min", "max", "std", "var"],
    "Age": ["mean", "count"]
})

multi_agg_data[("Salary", "std")] = multi_agg_data[("Salary", "std")].round(2)
multi_agg_data[("Salary", "var")] = multi_agg_data[("Salary", "var")].round(2)
multi_agg_data
```
