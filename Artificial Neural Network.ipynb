```python
n = 3
num_hidden_layers = 4
m = [2, 3, 1, 5]
num_output_nodes = 2
```

```python
import numpy as np

num_nodes_prev = n

network = {}

for layer in range(num_hidden_layers + 1):
    if layer == num_hidden_layers:
        layer_name = 'output'
        num_nodes = num_output_nodes
    else :
        layer_name = 'h_layer_{}'.format(layer+1)
        num_nodes = m[layer]

    network[layer_name] = {}
    for node in range(num_nodes):
        node_name = 'node_{}'.format(node+1)
        network[layer_name][node_name] = {
        'weights': np.around(np.random.uniform(size=num_nodes_prev), decimals=2),
        'biases': np.around(np.random.uniform(size=1), decimals=2)
        }
    num_nodes_prev = num_nodes

print(network)
```

```python
def initialize_network (num_inputs, num_hidden_layers, num_hidden_nodes, num_output_nodes):
    num_prev_node = num_inputs
    network = {}
    for layer in range(num_hidden_layers+1):
        if layer == num_hidden_layers:
            layer_name = 'output'
            num_node = num_output_nodes
        else :
            layer_name = 'layer_{}'.format(layer+1)
            num_node = num_hidden_nodes[layer]

        network[layer_name] = {}
        for node in range(num_node):
            node_name = 'node_{}'.format(node+1)
            network[layer_name][node_name] = {
                'weights': np.around(np.random.uniform(size=num_prev_node), decimals=2),
                'bias': np.around(np.random.uniform(size=1), decimals=2)
            }
        num_prev_node = num_node

    return network
```

```python
small_network = initialize_network(2, 4, [2, 5, 3, 7], 3)
```

```python
def compute_weighted_sum(inputs, weights, bias):
    return np.sum(inputs * weights) + bias
```

```python
from random import seed
import numpy as np

np.random.seed(5)
inputs = np.around(np.random.uniform(size=2), decimals=2)

print('The inputs to the network are {}'.format(inputs))
```

```python
node_weights = small_network['layer_1']['node_1']['weights']
node_bias = small_network['layer_1']['node_1']['bias']

compute_weighted_sum(inputs, node_weights, node_bias)
```

```python
def node_activation(weighted_sum):
    return 1.0 / (1.0 + np.exp(-1 * weighted_sum))
```

```python
np.around(node_activation(compute_weighted_sum(inputs, node_weights, node_bias))[0], decimals=4)
```

```python
def forward_propagate(network, inputs):
    layer_inputs = inputs
    for layer in network:
        layer_data = network[layer]

        layer_outputs = []
        for node in layer_data:
            node_data = layer_data[node]

            node_output = node_activation(compute_weighted_sum(layer_inputs, node_data['weights'], node_data['bias']))
            
            layer_outputs.append(float(np.around(node_output[0], decimals=4)))

        if layer != 'output':
            print('The outputs of the nodes in hidden layer number {} is {}'.format(layer.split('_')[1], layer_outputs))
    
        layer_inputs = layer_outputs

    network_predictions = layer_outputs
    return network_predictions
```

```python
forward_propagate(small_network, inputs)
```
