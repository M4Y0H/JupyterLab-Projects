```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import confusion_matrix
from datetime import datetime
```

```python
iris = load_iris()
features, classify = iris.data, iris.target
np.random.seed(42)
f_train, f_test, c_train, c_test = train_test_split(features, classify, test_size=0.2, stratify=classify)

pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('pca', PCA()),
    ('knn', KNeighborsClassifier())
])
model_params = {
    'pca__n_components': [2,3,4],
    'knn__n_neighbors': [3,4,5,6,7]
}
cv = StratifiedKFold(n_splits=5, shuffle=True)
initial =  datetime.now()
best_stats = GridSearchCV(estimator=pipeline, param_grid=model_params, cv=cv, scoring='accuracy', verbose=2, n_jobs=-1)
final = datetime.now()
time_taken = final - initial
display(time_taken)
```
